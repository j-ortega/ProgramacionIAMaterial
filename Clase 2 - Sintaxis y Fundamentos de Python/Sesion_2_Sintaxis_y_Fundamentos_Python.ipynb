## 2. Tipos de datos en Python

- **Numéricos:** `int`, `float`, `complex`

  
>- int (Enteros)
>>- Enteros de precisión arbitraria (limitados por la memoria disponible, no por el lenguaje).
>>- Literales en decimal y también binario (0b...), octal (0o...) y hexadecimal (0x...).
>>- Ejemplos:
```python
diez = 10
diez_bin = 0b1010
diez_oct = 0o12
diez_hex = 0xA
```

>- float (Punto flotante)
>>- Representación de cualquier número real.
>>- Al igual que los números enteros, un float se crea a partir de un literal, o bien como resultado de una expresión o una función.
>>- Ejemplos:
```python
un_real = 1.1  # El literal debe incluir el carácter .
otro_real = 1/2  # El resultado de 1/2 es un float
not_cient = 1.23E3  # float con notación científica (1230.0)
```

>- complex (Complejos)
>>- Los números complejos tienen una parte real y otra imaginaria y cada una de ellas se representa como un float.
>>- Para crear un número complejo, se sigue la siguiente estructura `<parte_real>+<parte_imaginaria>j`. Y se puede acceder a la parte real e imaginaria a través de los atributos `real` e `imag`:
>>- Ejemplos:
```python
z = 1 + 2j
z.real, z.imag  # (1.0, 2.0)
```

***Aritmética con tipos numéricos:***  
Con todos los tipos numéricos se pueden aplicar las operaciones de la aritmética: suma, resta, producto, división, …  
En Python está permitido realizar una operación aritmética con números de distinto tipo.  
En este caso, el tipo numérico «más pequeño» se convierte al del tipo «más grande», de manera que el tipo del resultado siempre es el del tipo mayor.  
Entendemos que el tipo int es menor que el tipo float que a su vez es menor que el tipo complex.

>- La principal característica de este tipo de datos es que es una colección cuyos elementos no guardan ningún orden y que además son únicos.


- **Mapeos:** `dict`


>- La clase `dict` de Python es un tipo mapa que asocia claves a valores.
>- A diferencia de los tipos secuenciales (list, tuple, range o str), que son indexados por un índice numérico, los diccionarios son indexados por claves. Estas claves siempre deben ser de un tipo inmutable, concretamente un tipo hashable.
>- 🎯 *NOTA*: Un objeto es hashable si tiene un valor de hash que no cambia durante todo su ciclo de vida. En principio, los objetos que son instancias de clases definidas por el usuario son hashables. También lo son la mayoría de tipos inmutables definidos por Python (int, float o str).
>- Pensad siempre en un diccionario como un contenedor de pares clave: valor, en el que la clave puede ser de cualquier tipo hashable y es única en el diccionario que la contiene. Generalmente, se suelen usar como claves los tipos `int` y `str`.
>- Las principales operaciones que se suelen realizar con diccionarios son almacenar un valor asociado a una clave y recuperar un valor a partir de una clave. Esta es la esencia de los diccionarios y es aquí donde son realmente importantes. En un diccionario, el acceso a un elemento a partir de una clave es una operación realmente rápida, eficaz y que consume pocos recursos si lo comparamos con cómo lo haríamos con otros tipos de datos.
>- Otras características a resaltar de los diccionarios:
>>- Es un tipo mutable, es decir, su contenido se puede modificar después de haber sido creado.
>>- Es un tipo ordenado. Preserva el orden en que se insertan los pares clave: valor.
>- Formas de crear un diccionario en Python:

```python
# 1. Pares clave: valor encerrados entre llaves
>>> d = {'uno': 1, 'dos': 2, 'tres': 3}
>>> d
{'uno': 1, 'dos': 2, 'tres': 3}

# 2. Argumentos con nombre
>>> d2 = dict(uno=1, dos=2, tres=3)
>>> d2
{'uno': 1, 'dos': 2, 'tres': 3}

# 3. Pares clave: valor encerrados entre llaves
>>> d3 = dict({'uno': 1, 'dos': 2, 'tres': 3})
>>> d3
{'uno': 1, 'dos': 2, 'tres': 3}

# 4. Iterable que contiene iterables con dos elementos
>>> d4 = dict([('uno', 1), ('dos', 2), ('tres', 3)])
>>> d4
{'uno': 1, 'dos': 2, 'tres': 3}

# 5. Diccionario vacío
>>> d5 = {}
>>> d5
{}

# 6. Diccionario vacío usando el constructor
>>> d6 = dict()
>>> d6
{}
```

**DOCUMENTACION OFICIAL:** https://docs.python.org/es/3/tutorial/datastructures.html#

````python
# Ejemplo: Inmutabilidad vs Mutabilidad en claves de diccionarios
# Las claves deben ser HASHABLES (inmutables).
# Una tupla es inmutable y sirve como clave:
coords = {(10, 20): "Punto A"}
print("Tupla como clave válida:", coords[(10, 20)])

# Pero si una tupla contiene un objeto mutable (como una lista), deja de ser hashable:
try:
    clave_invalida = (10, [1, 2])
    dic_error = {clave_invalida: "Error"}
except TypeError as e:
    print("Capturado TypeError esperado al usar tupla con lista como clave:")
    print(" ->", e)
````

### Mutabilidad y semántica de copia

**Objeto MUTABLE**:
>- Su estado interno puede modificarse “in place” (sin cambiar su identidad).
>- Ejemplos: `list`, `dict`, `set`.

**Objeto INMUTABLE**:
>- Una vez creado, no puede cambiar. Cualquier “modificación” crea otro objeto.
>- Ejemplos: `int`, `float`, `bool`, `str`, `tuple`, `frozenset`, `range`, `NoneType`.

````python
# Ejemplos básicos
n = 42              # int
x = 3.14            # float
flag = True         # bool
s = "Hola, Python"  # str
lista = [1, 2, 3]   # list (mutable)
tupla = (1, 2, 3)   # tuple (inmutable)
conjunto = {1, 2, 2, 3}  # set (sin duplicados)
dic = {"a": 1, "b": 2}   # dict

print(type(n), type(x), type(flag), type(s), type(lista), type(tupla), type(conjunto), type(dic))
print(n, x, flag, s, lista, tupla, conjunto, dic)
````

#### 1) Indexado y *slicing* en `str`, `list`, `tuple`

**Indexado (`obj[i]`)**

- Índices empiezan en `0`.
- Índices negativos cuentan desde el final (`-1` es el último).
- Lanza `IndexError` si el índice está fuera de rango.

```python
s = "Python"
lst = [10, 20, 30, 40]
t = ('a', 'b', 'c')

s[0], s[-1]      # ('P', 'n')
lst[2], lst[-2]  # (30, 30)
t[1]             # 'b'
```

**Slicing (`obj[inicio:fin:paso]`)**

- Toma desde `inicio` **incluido** hasta `fin` **excluido`**.
- Puedes omitir partes: `[:fin]`, `[inicio:]`, `[::paso]`.
- No falla si los límites se salen del rango.
- `paso` negativo recorre hacia atrás (útil para invertir).

```python
s = "Python"
s[1:4]     # 'yth'
s[:3]      # 'Pyt'
s[3:]      # 'hon'
s[::-1]    # 'nohtyP'  (invertir)
s[::2]     # 'Pto'     (uno sí, uno no)

lst = [10, 20, 30, 40, 50]
lst[1:4]   # [20, 30, 40]
lst[-4:-1] # [20, 30, 40]
```

## 3. Estructuras de control: condicionales y bucles

#### Condicionales: `if / elif / else`

Se utilizan para ejecutar código dependiendo de si una condición es **verdadera** o **falsa**.

>- `if`: evalúa la primera condición.
>- `elif`: evalúa condiciones adicionales si la anterior fue falsa.
>- `else`: se ejecuta si ninguna condición previa se cumple.
>- **IMPORTANTE:** La indentación es obligatoria y marca el bloque de código.

```python
x = 10

if x > 0:
    print("x es positivo")
elif x == 0:
    print("x es cero")
else:
    print("x es negativo")
```

#### Bucles `for` sobre secuencias/iterables

Se usa para **iterar** sobre una colección (lista, cadena, rango, etc.).

>- range(n) genera una secuencia de 0 a n-1.
>- for en Python itera directamente sobre los elementos, no sobre índices (aunque puedes usar `enumerate` si necesitas ambos).

```python
frutas = ["manzana", "plátano", "cereza"]

for fruta in frutas:
    print("Me gusta la", fruta)

# Ejemplo con range
for i in range(5):
    print("Iteración:", i)
```

#### Bucles `while`

Se utiliza cuando no sabemos de antemano cuántas iteraciones habrá, y queremos repetir **mientras se cumpla una condición**.

>- Es importante actualizar la variable de control dentro del bucle. Si no lo haces, corres riesgo de un bucle infinito.

```python
contador = 0

while contador < 5:
    print("Contador:", contador)
    contador += 1
```

#### Control de bucles: `break` y `continue`

- **`break`**: interrumpe el bucle inmediatamente.  
- **`continue`**: salta a la siguiente iteración, sin ejecutar el resto del bloque.

```python
# break
for numero in range(10):
    if numero == 5:
        break
    print(numero)

# continue
for numero in range(5):
    if numero == 2:
        continue
    print(numero)
```

#### Bucles con `else`

En Python, los bucles `for` y `while` pueden llevar un `else`.  
El bloque `else` se ejecuta **solo si el bucle terminó sin usar `break`**.

```python
for numero in range(3):
    print("Número:", numero)
else:
    print("Bucle terminado sin break")

# Con break
for numero in range(3):
    if numero == 1:
        break
    print("Número:", numero)
else:
    print("Esto no se imprime porque hubo break")
```

## 4. Funciones y parámetros; `lambda`, `map`, `filter`, `zip`, `enumerate`

> Una función es una forma de encapsular un conjunto de instrucciones que se pueden reutilizar tantas veces como sea necesario dentro de un programa.  
Gracias a las funciones, el código se vuelve más organizado, modular y sigue el principio DRY (Don’t Repeat Yourself), evitando repetir lógica en diferentes partes del proyecto.  
Python incluye numerosas funciones incorporadas como print() o len(), pero también es posible definir funciones propias adaptadas a las necesidades de cada aplicación.  
Un beneficio clave de emplear funciones es que simplifican el código, disminuyendo la cantidad de líneas y mejorando su legibilidad.  
Las variables definidas dentro de una función solo existen dentro del ámbito de esa función.

> Sintaxis:
>> La palabra reservada def  
>> Un identificador que servirá como nombre de la función  
>> Paréntesis (), que pueden incluir parámetros de entrada (opcionales)  
>> Dos puntos : al final de la cabecera  
>> Un bloque de instrucciones con sangría que se ejecutará al llamar a la función  
>> Una sentencia return (opcional), para devolver un valor  

```python
def saludar(nombre):
    mensaje = f"Hola, {nombre}!"
    return mensaje  # return es opcional, pero aquí devuelve el saludo

# Llamada a la función
print(saludar("Ana"))
print(saludar("Luis"))
```

#### 1) Parámetros por **posición** y por **nombre**

Al llamar a una función, los argumentos se pueden pasar:

- **Por posición**: en el orden en que están definidos.
- **Por nombre (keywords)**: especificando explícitamente el nombre del parámetro.

#### 2) Valores por defecto

Podemos asignar valores por defecto a los parámetros.  
⚠️ Cuidado con usar **objetos mutables** (listas, diccionarios, sets) como valores por defecto.

````python
def potencia(base, exponente=2):
    return base ** exponente

print(potencia(3))   # usa el valor por defecto 2
print(potencia(3, 3))

# Problema de mutables
def agregar_item(item, lista=[]):
    lista.append(item)
    return lista

print(agregar_item("a"))
print(agregar_item("b"))

# Solución
def agregar_item_seguro(item, lista=None):
    if lista is None:
        lista = []
    lista.append(item)
    return lista

print(agregar_item_seguro("a"))
print(agregar_item_seguro("b"))
````

#### 3) Argumentos variables: `*args` y `**kwargs`

- `*args`: permite pasar un número variable de argumentos posicionales.
- `**kwargs`: permite pasar un número variable de argumentos nombrados (diccionario).

````python
def suma(*args):
    print(type(args))
    return sum(args)

print(suma(1, 2, 3, 4))

def mostrar_info(nombre, edad, **kwargs):
    for clave, valor in kwargs.items():
        print(f"{clave}: {valor}")

mostrar_info(nombre="Ana", edad=25, ciudad="Madrid")
````

#### 4) Funciones `map`, `lambda`, `filter`, `zip`, `enumerate`


>- **`map`**: aplica una función a cada elemento
```python
nums = [1, 2, 3, 4, 5]
print(list(map(lambda x: x*2, nums)))
```
>- Una **lambda** es una función anónima, útil para operaciones simples y rápidas.
```python
cuadrado = lambda x: x**2
print(cuadrado(5))

# Uso en listas
nums = [1, 2, 3, 4, 5]
print(list(map(lambda x: x**2, nums)))
```
>- **`filter`**: filtra según una condición
```python
pares = list(filter(lambda x: x % 2 == 0, nums))
print(pares)
```
>- **`zip`**: combina secuencias en pares
```python
nombres = ["Ana", "Luis", "Marta"]
edades = [23, 30, 19]
```

## 5. Tipado estático con `typing`

Python es un lenguaje de **tipado dinámico**, lo que significa que no es obligatorio indicar el tipo de las variables ni de los parámetros de una función.  
Sin embargo, a partir de **Python 3.5** se introdujo el módulo `typing`, que permite **anotar tipos** para mejorar la legibilidad del código y facilitar el uso de herramientas como linters o analizadores estáticos (por ejemplo, `mypy`).

👉 Estas anotaciones **no cambian el comportamiento** del programa en tiempo de ejecución, solo sirven como documentación y para comprobaciones externas.

- `List[int]`, `Dict[str, float]`, `Tuple[int, str]`, `Set[str]`
- `Optional[T]` (equivale a `Union[T, None]`)
- `Union[A, B]`
- `Literal["a", "b"]` para valores concretos
- `Callable[[A, B], R]` para funciones

## 6. Docstrings: formato, utilidad, estándares

Un **docstring** es una cadena de documentación que se escribe al inicio de un módulo,
clase o función para describir qué hace, cómo funciona y cómo debe usarse.  
Los docstrings sirven como **documentación interna** del código y son accesibles
mediante la función `help()` o el atributo `.__doc__`.

#### 1) Utilidad de los docstrings
- Facilitan la comprensión del código.
- Sirven como documentación oficial de librerías y módulos.
- Algunas herramientas (como Sphinx o pydoc) pueden generar documentación automática a partir de los docstrings.
- Ayudan a mantener el código legible y profesional

#### 2) Sintaxis básica

Un docstring se escribe entre triples comillas (`""" ... """` o `''' ... '''`) y suele colocarse justo después de la definición de una función, clase o módulo.

````python
def suma(a, b):
    """
    Devuelve la suma de dos números.

    Args:
        a (int): primer número.
        b (int): segundo número.

    Returns:
        int: la suma de `a` y `b`.
    """
    return a + b

help(suma)
````

#### 3) Google Style

La convención de Google es una de las más populares para escribir docstrings en Python.
Se estructura de la siguiente manera:

- **Descripción breve** de lo que hace la función.
- Sección `Args:` con la lista de parámetros.
- Sección `Returns:` con lo que devuelve la función.
- Sección `Raises:` con las excepciones que la función puede lanzar intencionadamente.

```python
def dividir(dividendo: float, divisor: float) -> float:
    """Calcula la división entre dos números.

    Args:
        dividendo (float): Número que se divide.
        divisor (float): Número por el cual se divide.

    Returns:
        float: Resultado de la división.

    Raises:
        ValueError: Si divisor es igual a cero.
    """
    if divisor == 0:
        raise ValueError("El divisor no puede ser cero.")
    return dividendo / divisor
```

Ejemplo completo: https://sphinxcontrib-napoleon.readthedocs.io/en/latest/example_google.html

````python
def multiplicar(a: int, b: int) -> int:
    """
    Multiplica dos números.

    Args:
        a (int): Primer número.
        b (int): Segundo número.

    Returns:
        int: El producto de `a` y `b`.
    """
    return a * b
````

## 8. Manejo de errores y excepciones

Bloques `try / except / else / finally`:
- `try`: código que puede fallar
- `except`: manejo del error
- `else`: se ejecuta si **no** hubo excepción
- `finally`: se ejecuta **siempre** (haya o no excepción)

Esto permite capturar y controlar errores sin que el programa se detenga de forma abrupta.

#### `try / except`

````python
try:
    f = open("no_existe.txt", "r", encoding="utf-8")
    contenido = f.read()
except Exception as e:
    print("⚠️ Error:", e)
````

#### Uso de `else` (se ejecuta si no hay error)

````python
try:
    f = open(os.path.join(base, "poema.txt"), "r", encoding="utf-8")
except FileNotFoundError:
    print("Archivo no encontrado")
else:
    print("Archivo abierto correctamente")
    f.close()
````

````python
# Demostración del flujo con 'else' en try/except:
# El bloque 'else' se ejecuta ÚNICAMENTE si el 'try' no arrojó ninguna excepción
def calcular_inverso(n):
    try:
        resultado = 100 / n
    except ZeroDivisionError:
        print(" -> Bloque except: División por cero no permitida.")
    else:
        print(f" -> Bloque else: Operación exitosa, resultado = {resultado}")
    finally:
        print(" -> Bloque finally: Siempre se ejecuta.")

print("Caso 1: Con error")
calcular_inverso(0)

print("\nCaso 2: Sin error")
calcular_inverso(4)
````

#### Uso de `finally` (se ejecuta siempre)

````python
try:
    f = open(os.path.join(base, "poema.txt"), "r", encoding="utf-8")
    print(f.readline())
except Exception as e:
    print("Error:", e)
else:
    print("Lectura exitosa")
finally:
    print("Bloque finally ejecutado siempre")
````

#### Captura múltiple de excepciones

````python
def dividir(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("No se puede dividir entre cero")
        return "HOLA"
    except TypeError:
        print("Los argumentos deben ser números")

print(dividir(10, 2))
print(dividir(5, 0))
print(dividir("a", 2))
````

#### Captura genérica con `Exception`

````python
try:
    x = int("abc")  # provoca ValueError
except Exception as e:
    print("Error capturado:", type(e).__name__, "-", e)
````

#### Lanzar excepciones manualmente con `raise`

````python
def raiz_cuadrada(x):
    if x < 0:
        raise ValueError("No se puede calcular raíz de número negativo")
    return x ** 0.5

print(raiz_cuadrada(9))
print(raiz_cuadrada(-4))  # genera excepción
````

## 9. La estructura main:  `if __name__ == "__main__":`:

¿Qué es `__name__`?  
Cuando Python ejecuta un archivo:
- Si se ejecuta **directamente** (`python archivo.py`), entonces `__name__ == "__main__"`.
- Si el archivo se **importa** como módulo (`import archivo`), entonces `__name__ == "nombre_del_módulo"`.

````python
__name__
````

#### Patrón canónico

````python
def main():
    print("Ejecutando lógica principal...")

if __name__ == "__main__":
    print("HOLA")
    main()
````

> Al usar este patrón, el código dentro del `if` **solo** corre cuando el archivo
se ejecuta directamente (no cuando es importado por otro).

````python
# Creamos un módulo temporal en el directorio de trabajo
from pathlib import Path

Path("mod_demo.py").write_text(
    """
print("[mod_demo] Cargando el módulo... __name__ =", __name__)

def main():
    print("[mod_demo] Dentro de main()")

if __name__ == "__main__":
    print("[mod_demo] Ejecutado como script")
    main()
else:
    print("[mod_demo] Importado como módulo")
""",
    encoding="utf-8"
)

print("Archivo 'mod_demo.py' creado.")
````

````python
# Ahora importamos mod_demo: como es una importación, __name__ dentro de mod_demo será 'mod_demo'
import importlib
import sys

if "mod_demo" in sys.modules:
    del sys.modules["mod_demo"]  # aseguramos recarga limpia

importlib.import_module("mod_demo")
````

> Observa en la salida que `mod_demo` detecta que ha sido **importado** y por tanto
no ejecuta el bloque protegido por `if __name__ == "__main__":`.

## 10. Comprehensions: list, dict y set

Sintaxis compacta y legible:

- **List:** `[expr for x in xs if cond]`
- **Dict:** `{k_expr: v_expr for x in xs if cond]`
- **Set:** `{expr for x in xs if cond]`
- **Generator:** `(expr for x in xs)`


#### 1) List Comprehensions

````python
# Ejemplo: cuadrados de 0 a 9
cuadrados = [x**2 for x in range(10)]
cuadrados
````

````python
# Con condición: solo pares
pares = [x for x in range(10) if x % 2 == 0]
pares
````

````python
pares = []
for x in range(10):
    if x % 2 == 0:
        pares.append(x)

pares
````

````python
cuadrados = []
for x in range(10):
    cuadrados.append(x**2)
cuadrados
````

#### 2) Dict Comprehensions

````python
# Ejemplo: diccionario de números y su cuadrado
dic = {x: x**2 for x in range(5)}
dic
````

````python
# Con condición: solo impares
dic_impares = {x: x**2 for x in range(10) if x % 2 == 1}
dic_impares
````

#### 3) Set Comprehensions

````python
# Ejemplo: conjunto de caracteres únicos en una cadena
texto = "abracadabra"
unicos = {c for c in texto}
unicos
````

````python
# Con condición: solo vocales
vocales = {c for c in texto if c in "aeiou"}
vocales
````

#### 4) Comprehensions anidados
También pueden anidarse para recorrer estructuras 2D.

````python
# Ejemplo: lista de pares (x, y)
pares = [(x, y) for x in range(3) for y in range(2)]
pares
````

````python
# Ejemplo: matriz 3x3 con comprensión anidada
matriz = [[x*y for y in range(3)] for x in range(3)]
matriz
````

#### 5) Buenas prácticas
- Úsalas cuando la lógica sea **simple y clara**.
- Si la expresión es compleja, mejor un bucle explícito.
- Evita comprensiones demasiado anidadas (difícil de leer).
